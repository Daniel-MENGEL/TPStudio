Il n’est pas demandé de reformuler ces objectifs généraux. Pour chaque nouvelle manipulation, vous préciserez en revanche son objectif propre avant de décrire le protocole.

```python
import numpy as np

N = 10_000

def degmin2deg(degres, minutes):
    return degres + minutes / 60

def ecart_angulaire(angle1, angle2):
    ecart = abs(angle2 - angle1)
    return np.minimum(ecart, 360 - ecart)

def ecart_normalise(x, ux, x_ref, ux_ref=0):
    return abs(x - x_ref) / np.sqrt(ux**2 + ux_ref**2)
```

## 2. Mesure de l’angle au sommet du prisme

La lunette autocollimatrice permet de repérer successivement les normales aux deux faces du prisme. On relève leurs positions angulaires $\alpha_1$ et $\alpha_2$ au vernier.

En tenant compte de la périodicité des angles, si $\Delta\alpha$ est le plus petit écart entre les deux positions :

$$A=180^\circ-\Delta\alpha$$

La valeur constructeur attendue est $A_{\mathrm{ref}}=60^\circ$.

### Objectif et protocole de mesure de $A$

Indiquez l’objectif propre de cette manipulation. Décrivez le positionnement du prisme, la manière de viser successivement les deux faces, les lectures au vernier et le choix des demi-largeurs d’intervalle associé à chaque lecture.

### Réponse :

L’objectif est de mesurer l’angle au sommet du prisme. Nous avons placé successivement la lunette en autocollimation sur chacune des deux faces et relevé les positions alpha1 et alpha2 au vernier. Pour chaque lecture, la demi-largeur retenue correspond à l’intervalle dans lequel la coïncidence entre le réticule et son image restait acceptable.

</div>

```python
# Lectures du binôme au vernier : degrés, minutes et demi-largeur en minutes
alpha1_deg = 91
alpha1_min = 72
Delta_alpha1_min = 2
alpha2_deg = 212
alpha2_min = 12
Delta_alpha2_min = 2

assert None not in [alpha1_deg, alpha1_min, Delta_alpha1_min,
                    alpha2_deg, alpha2_min, Delta_alpha2_min],     "Compléter les deux lectures et leurs demi-largeurs."

alpha1_m = degmin2deg(alpha1_deg, alpha1_min)
alpha2_m = degmin2deg(alpha2_deg, alpha2_min)
Delta_alpha1 = Delta_alpha1_min / 60
Delta_alpha2 = Delta_alpha2_min / 60

alpha1 = np.random.uniform(alpha1_m - Delta_alpha1, alpha1_m + Delta_alpha1, N)
alpha2 = np.random.uniform(alpha2_m - Delta_alpha2, alpha2_m + Delta_alpha2, N)

Delta_alpha = ecart_angulaire(alpha1, alpha2)
A_values = 180 - Delta_alpha
A = A_values.mean()
u_A = A_values.std()
A_constructeur = 60
E_n_A = ecart_normalise(A, u_A, A_constructeur)

print("A =", round(A, 2), "+/-", round(u_A, 2), "°")
print("Valeur constructeur =", round(A_constructeur, 1), "°")
print("Écart normalisé =", round(E_n_A, 2))
```

## 3. Minimum de déviation et indice du prisme

Éclairez le prisme avec la lampe à vapeur de sodium. Recherchez le minimum de déviation d’un côté, relevez la position $\beta_1$ de la lunette, puis faites pivoter le prisme d’environ $180^\circ$ et recommencez de l’autre côté pour obtenir $\beta_2$.

$$D_m=\frac{1}{2}\,\Delta\beta$$

où $\Delta\beta$ est le plus petit écart angulaire entre les deux positions. L’indice est ensuite donné par :

$$n=\frac{\sin\left(\frac{A+D_m}{2}\right)}{\sin\left(\frac{A}{2}\right)}$$

Les angles doivent être convertis en radians dans les fonctions trigonométriques de Python.

### Objectif et protocole de mesure de $D_m$

Indiquez l’objectif propre de cette manipulation. Décrivez comment vous repérez le changement de sens de déplacement de l’image, affinez chaque minimum, réalisez les deux lectures symétriques et estimez leurs demi-largeurs d’intervalle.

### Réponse :

L’objectif est de mesurer le minimum de déviation afin d’en déduire l’indice du prisme. Pour chaque orientation du prisme, nous avons suivi l’image jaune pendant la rotation du plateau et repéré la position où son déplacement changeait de sens. Après blocage, nous avons affiné la coïncidence avec le fil vertical du réticule et relevé beta1 puis beta2. Les demi-largeurs décrivent l’intervalle où le minimum restait indiscernable.

</div>

```python
# Deux positions symétriques du minimum de déviation
beta1_deg = 40
beta1_min = 10
Delta_beta1_min = 3
beta2_deg = 166
beta2_min = 44
Delta_beta2_min = 3

assert None not in [beta1_deg, beta1_min, Delta_beta1_min,
                    beta2_deg, beta2_min, Delta_beta2_min],     "Compléter les deux positions du minimum et leurs demi-largeurs."

beta1_m = degmin2deg(beta1_deg, beta1_min)
beta2_m = degmin2deg(beta2_deg, beta2_min)
Delta_beta1 = Delta_beta1_min / 60
Delta_beta2 = Delta_beta2_min / 60

beta1 = np.random.uniform(beta1_m - Delta_beta1, beta1_m + Delta_beta1, N)
beta2 = np.random.uniform(beta2_m - Delta_beta2, beta2_m + Delta_beta2, N)

Dm_values = ecart_angulaire(beta1, beta2) / 2
Dm = Dm_values.mean()
u_Dm = Dm_values.std()

A_rad = np.deg2rad(A_values)
Dm_rad = np.deg2rad(Dm_values)
n_values = np.sin((A_rad + Dm_rad) / 2) / np.sin(A_rad / 2)
n = n_values.mean()
u_n = n_values.std()
n_constructeur = 1.76
u_n_constructeur = 0.01
E_n_n = ecart_normalise(n, u_n, n_constructeur, u_n_constructeur)

print("D_m =", round(Dm, 2), "+/-", round(u_Dm, 2), "°")
print("n =", round(n, 4), "+/-", round(u_n, 4))
print("Valeur constructeur =", round(n_constructeur, 2), "+/-", round(u_n_constructeur, 2))
print("Écart normalisé =", round(E_n_n, 2))
```
