## 3. Relation de conjugaison

La relation de Descartes s'écrit :

$$\frac{1}{\overline{OA'}}-\frac{1}{\overline{OA}}=\frac{1}{f'}$$

avec $\overline{OA}=d_0-d_1$ et $\overline{OA'}=d_2-d_1$. Les positions sont exprimées en centimètres dans ce notebook.

### 3.1 Mesure unique avec incertitude de type B

Relevez $d_0$, $d_1$ et $d_2$. Pour chaque position, estimez un intervalle raisonnable $[d-\Delta d,d+\Delta d]$. La simulation utilise une loi uniforme avec `np.random.uniform`, conformément au TP-cours Incertitudes.

<!-- single-uncertainty-response -->
<div style="background-color:#F5F8FC; border:1px solid #DCE6F0; padding:12px 16px; border-radius:4px; color:#202124;">

### Justification des incertitudes de position

Indiquez l'objectif de cette manipulation (en relisant l'énoncé du TP si nécessaire), puis justifiez séparément les demi-largeurs $\Delta d_0$, $\Delta d_1$ et $\Delta d_2$ à partir des graduations, du repérage des éléments et de la plage de netteté acceptable.

### Réponse :

À compléter.

</div>

```python
# Mesures du binôme, en cm
d0m = None
Delta_d0 = None
d1m = None
Delta_d1 = None
d2m = None
Delta_d2 = None

assert None not in [d0m, Delta_d0, d1m, Delta_d1, d2m, Delta_d2], \
    "Compléter les positions et les demi-largeurs."

N = 50000
d0_samples = np.random.uniform(d0m - Delta_d0, d0m + Delta_d0, N)
d1_samples = np.random.uniform(d1m - Delta_d1, d1m + Delta_d1, N)
d2_samples = np.random.uniform(d2m - Delta_d2, d2m + Delta_d2, N)

OA_samples = d0_samples - d1_samples
OAp_samples = d2_samples - d1_samples
f_single_samples = OA_samples * OAp_samples / (OA_samples - OAp_samples)

f1 = f_single_samples.mean()
uf1 = f_single_samples.std(ddof=1)

decimales_f1 = None

assert decimales_f1 is not None, "Choisir le nombre de décimales à conserver dans le résultat."
print("f' par mesure unique =", round(f1, decimales_f1), "+/-", round(uf1, decimales_f1), "cm")
```

```python
# Comparaison à la valeur attendue pour la lentille +3,3 dioptries
f_th = 100 / 3.3
u_fth = 1.0
En_single_theory = abs(f1 - f_th) / np.sqrt(uf1**2 + u_fth**2)

print("Valeur attendue =", round(f_th, 1), "+/-", round(u_fth, 1), "cm")
print("Écart normalisé =", round(En_single_theory, 2))
```

### 3.2 Mesures multiples et validation de la relation de Descartes

Tant que la distance objet–écran est supérieure à $4f'$, deux positions de la lentille donnent une image nette. Pour plusieurs positions de l'écran, relevez environ dix couples $(d_1,d_2)$, en recherchant les deux positions de netteté.

On tracera $1/\overline{OA'}$ en fonction de $1/\overline{OA}$. La pente attendue = 1 et l'ordonnée à l'origine vaut $1/f'$.

<!-- multiple-protocol-response -->

```python
# Vérification de la relation de conjugaison
# Remplacer les ? et les points de suspension par toutes les mesures du binôme, en cm.
d0 = None
d1 = np.array([?, ..., ?], dtype=float)
d2 = np.array([?, ..., ?], dtype=float)

assert d0 is not None, "Compléter d0."
assert len(d1) == len(d2), "Les tableaux d1 et d2 doivent avoir la même longueur."
assert len(d1) >= 10, "On attend environ une dizaine de couples de mesures."

OA = d0 - d1
OAp = d2 - d1
invOA = 1 / OA
invOAp = 1 / OAp
a, b = np.polyfit(invOA, invOAp, 1)

plt.figure()
plt.plot(invOA, invOAp, "bo", label="Mesures")
plt.plot(invOA, a * invOA + b, "r-", label="Régression affine")
plt.xlabel(r"$1/\overline{OA}$ en cm$^{-1}$")
plt.ylabel(r"$1/\overline{OA'}$ en cm$^{-1}$")
plt.title("Validation de la relation de conjugaison")
plt.legend()
plt.show()

print("Pente a =", round(a, 3))
print("Ordonnée à l'origine b =", round(b, 4), "cm^-1")
print("Distance focale déduite de b =", round(1 / b, 1), "cm")
```

```python
# Détermination statistique de f' pour chaque couple
f_multiple = OA * OAp / (OA - OAp)
f2 = f_multiple.mean()
uf2 = f_multiple.std(ddof=1)

En_multiple_theory = abs(f2 - f_th) / np.sqrt(uf2**2 + u_fth**2)
En_single_multiple = abs(f2 - f1) / np.sqrt(uf2**2 + uf1**2)

decimales_f2 = None

assert decimales_f2 is not None, "Choisir le nombre de décimales à conserver dans le résultat."
print("Valeurs de f' =", np.round(f_multiple, decimales_f2), "cm")
print("f' moyen =", round(f2, decimales_f2), "+/-", round(uf2, decimales_f2), "cm")
print("Écart normalisé avec la valeur attendue =", round(En_multiple_theory, 2))
print("Écart normalisé entre les deux méthodes =", round(En_single_multiple, 2))
```
