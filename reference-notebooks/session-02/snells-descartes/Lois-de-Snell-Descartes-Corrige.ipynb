## 1. Première méthode : angle limite

On étudie le passage Plexiglas $\to$ air. À la limite de réflexion totale :

$$\sin i_l = \frac{1}{n} \qquad\text{donc}\qquad n = \frac{1}{\sin i_l}$$

<!-- critical-protocol-response -->
<div style="background-color:#F5F8FC; border:1px solid #DCE6F0; padding:12px 16px; border-radius:4px; color:#202124;">

### Objectif et protocole de mesure de l'angle limite

Indiquez l'objectif de cette manipulation (en relisant l'énoncé du TP si nécessaire), puis décrivez comment vous repérez l'apparition de la réflexion totale, mesurez $i_l$ et estimez l'incertitude-type $u(i_l)$. Mentionnez les principales difficultés de lecture.

### Réponse :

L'objectif est de déterminer l'indice du Plexiglas à partir de l'angle limite. On place le montage dans le sens Plexiglas vers air puis on augmente progressivement l'incidence. L'angle limite est relevé lorsque le rayon transmis disparaît et que la réflexion totale apparaît. Plusieurs approches de part et d'autre de la transition permettent d'encadrer $i_l$. La graduation et le caractère progressif de la disparition conduisent ici à $u(i_l)=1°$ à $2°$.

</div>

```python
# Mesures du binôme, en degrés
il_deg = 42
uil_deg = 1

assert il_deg is not None and uil_deg is not None, "Compléter il_deg et uil_deg."

# Conversion en radian
il=deg2rad(il_deg)
uil=deg2rad(uil_deg)

N = 50000
il_samples = np.random.normal(il,uil, N)
n_direct_samples = 1 / np.sin(il_samples)

n1 = n_direct_samples.mean()
un1 = n_direct_samples.std(ddof=1)

# Choisir le nombre de décimales conformément au TP-cours Incertitudes.
decimales_n1 = 2
print("n par angle limite =", round(n1, decimales_n1), "+/-", round(un1, decimales_n1))
```

Écrivez le résultat avec une incertitude correctement arrondie, puis commentez sa précision, les difficultés expérimentales et sa cohérence avec une valeur documentée pour le Plexiglas.

### Réponse :

On obtient un indice voisin de $1,50$ avec une incertitude de l'ordre de $0,03$. La méthode est assez précise, mais le repérage exact de la disparition du rayon réfracté reste délicat. Le résultat est cohérent avec les valeurs usuelles de l'indice du Plexiglas, proches de 1,49.

## 2. Seconde méthode : un couple d'angles

Pour une réfraction air $\to$ Plexiglas :

$$n = \frac{\sin i_1}{\sin i_2}$$

<!-- single-pair-protocol-response -->
<div style="background-color:#F5F8FC; border:1px solid #DCE6F0; padding:12px 16px; border-radius:4px; color:#202124;">

### Objectif et protocole avec un couple d'angles

Indiquez l'objectif de cette manipulation (en relisant l'énoncé du TP si nécessaire), puis décrivez la mesure de $i_1$ et $i_2$. Justifiez le choix d'un angle d'incidence « bien choisi » et les incertitudes-types retenues.

### Réponse :

L'objectif est de déterminer l'indice avec un seul couple d'angles et de comparer ce résultat à la première méthode. On replace le montage dans le sens air vers Plexiglas. On choisit $i_1=50,0°$ : l'angle est assez grand pour limiter l'effet relatif des incertitudes, sans rendre la lecture ambiguë. On mesure alors $i_2=31°$. La demi-graduation et la largeur des rayons conduisent à retenir $u(i_1)=u(i_2)=1°$.

</div>

```python

# Mesures du binôme, en degrés
i1_deg = 50.0
ui1_deg = 1
i2_deg = 31
ui2_deg = 1

assert None not in [i1_deg, ui1_deg, i2_deg, ui2_deg], "Compléter les quatre valeurs."

# Conversion en radian
i1=deg2rad(i1_deg)
ui1=deg2rad(ui1_deg)
i2=deg2rad(i2_deg)
ui2=deg2rad(ui2_deg)


N = 50000
i1_samples = np.random.normal(i1, ui1, N)
i2_samples = np.random.normal(i2, ui2, N)
n_geometric_samples = np.sin(i1_samples) / np.sin(i2_samples)

n2 = n_geometric_samples.mean()
un2 = n_geometric_samples.std(ddof=1)
En12 = abs(n2 - n1) / np.sqrt(un2**2 + un1**2)

decimales_n2 = 2
print("n par un couple d'angles =", round(n2, decimales_n2), "+/-", round(un2, decimales_n2))
print("Écart normalisé entre les méthodes 1 et 2 =", round(En12, 2))
```

### Réponse :

La seconde méthode donne un indice voisin de $1,49$ avec une incertitude d'environ $0,05$. L'écart normalisé entre les deux premières méthodes est inférieur à 2 : les résultats sont compatibles. La lecture de seulement deux angles explique que cette méthode soit moins précise.

## 3. Vérification graphique de la loi de la réfraction

Relevez **au moins quinze couples** $(i_1,i_2)$. Si la loi de Snell-Descartes est vérifiée pour air $\to$ Plexiglas, les points représentant $\sin i_1$ en fonction de $\sin i_2$ doivent suivre :

$$\sin i_1 = n\sin i_2.$$

La pente de la droite donne alors une estimation de $n$.

<!-- series-protocol-response -->
<div style="background-color:#F5F8FC; border:1px solid #DCE6F0; padding:12px 16px; border-radius:4px; color:#202124;">

### Objectif et protocole de la série angulaire

Indiquez l'objectif de cette manipulation (en relisant l'énoncé du TP si nécessaire), puis expliquez comment vous choisissez et relevez au moins quinze couples d'angles.

### Réponse :

L'objectif est de vérifier graphiquement la loi de Snell-Descartes et d'obtenir une troisième estimation de l'indice. On fait varier l'incidence de 0° à 70° et on relève au moins quinze couples $(i_1,i_2)$ répartis sur toute la plage utile.

</div>

```python
# Vérification de la loi de la réfraction
# Vérification graphique
# Affichage de l'équation de la droite
# Remplacer les ? et les points de suspension par toutes les mesures du binôme.
i1_deg_tableau = np.array([0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60, 65, 70], dtype=float)
i2_deg_tableau = np.array([0, 3.4, 6.7, 10.0, 13.0, 16.1, 19.2, 22.3, 25.3, 28.2, 30.9, 33.5, 35.9, 38.1, 39.1], dtype=float)

assert len(i1_deg_tableau) == len(i2_deg_tableau), "Les deux listes doivent avoir la même longueur."
assert len(i1_deg_tableau) >= 15, "Il faut au moins quinze couples de valeurs."


# Conversion en radian
i1_tableau=np.radians(i1_deg_tableau)
i2_tableau=np.radians(i2_deg_tableau)


sini1 = np.sin(i1_tableau)
sini2 = np.sin(i2_tableau)
a, b = np.polyfit(sini2, sini1, 1)

plt.figure()
plt.plot(sini2, sini1, "bo", label="Mesures")
plt.plot(sini2, a * sini2 + b, "r-", label="Régression affine")
plt.xlabel(r"$\sin i_2$")
plt.ylabel(r"$\sin i_1$")
plt.title("Vérification graphique de la loi de Snell-Descartes")
plt.legend()
plt.show()

print("Pente a =", round(a, 3))
print("Ordonnée à l'origine b =", round(b, 3))
```

<!-- graph-analysis-response -->
<div style="background-color:#F5F8FC; border:1px solid #DCE6F0; padding:12px 16px; border-radius:4px; color:#202124;">

### Analyse du graphe

Les points sont-ils alignés et l'ordonnée à l'origine est-elle compatible avec zéro ? Concluez sur la vérification de la loi de Snell-Descartes dans le domaine étudié, puis interprétez la pente.

### Réponse :

Les points représentant $\sin i_1$ en fonction de $\sin i_2$ sont alignés et l'ordonnée à l'origine est très proche de zéro. La loi de Snell-Descartes est donc vérifiée dans le domaine étudié. La pente de la droite représente l'indice du Plexiglas et vaut environ 1,48.

</div>

```python
# Méthode statistique
# Détermination de n pour chaque couple non nul
#Méthode statistique

sini1=np.delete(sini1,0) #pour éviter la divison par zéro
sini2=np.delete(sini2,0)

n_series = sini1/sini2
n3 = n_series.mean()
un3 = n_series.std(ddof=1)
En23 = abs(n3 - n2) / np.sqrt(un3**2 + un2**2)

decimales_n3 = 2
print("Valeurs de n =", np.round(n_series, decimales_n3))
print("n moyen =", round(n3, decimales_n3), "+/-", round(un3, decimales_n3))
print("Écart normalisé entre les méthodes 2 et 3 =", round(En23, 2))
```
